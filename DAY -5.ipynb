```python
# For numerical columns, we can use mean and median
# Example column: 'Age' (already complete, but showing the process)
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed'] = df['Age'].fillna(df['Age'].median())

# For categorical columns, use mode
# Example column: 'Cabin'
df['Cabin_mode_imputed'] = df['Cabin'].fillna(df['Cabin'].mode()[0])

# Display the resulting dataset with the imputed columns
df[['Age', 'Age_mean_imputed', 'Age_median_imputed', 'Cabin', 'Cabin_mode_imputed']]
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("titanic uncleaned.csv")

# Important numerical columhs
cols = ['Age', 'SibSp', 'Parch', 'Fare']

# Boxplots before removing outliers
plt.figure(figsize=(12, 8))
for i, col in enumerate(cols, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot Before Removing Outliers - {col}')

plt.tight_layout()
plt.show()
```

```python
#Best for Skewed Data

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)


IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]
print(f"Total outliers detected using IQR: {len(outliers_iqr)}")
print(outliers_iqr[["PassengerId", "Fare"]].head())
```

```python
from sklearn.preprocessing import LabelEncoder
file_path = "titanic uncleaned.csv"
titanic_data = pd.read_csv(file_path)
encoded_data = titanic_data.copy()
label_encoder = LabelEncoder()
encoded_data['Sex_Label'] = label_encoder.fit_transform(encoded_data['Sex'])
one_hot_encoded_data = pd.get_dummies(encoded_data['Embarked'], prefix='Embarked')
encoded_data = pd.concat([encoded_data, one_hot_encoded_data], axis=1)
pclass_mapping = {1: 'First', 2: 'Second', 3: 'Third'}
encoded_data['Pclass_Category'] = encoded_data['Pclass'].map(pclass_mapping)
print(encoded_data[['Sex','Sex_Label','Embarked','Embarked_Q','Embarked_S','Embarked_C','Pclass','Pclass_Category']].head())
```
